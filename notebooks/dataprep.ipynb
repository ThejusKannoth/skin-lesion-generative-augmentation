```python
#clean data by placing into foledrs for image
import os
import shutil
import pandas as pd
```

```python
metadata_path = "./dataverse_files/HAM10000_metadata"
images_path   = "./dataverse_files/HAM10000/"
output_path   = "./dataverse_files/HAM10000/Synthetic/"
```

Created class folders so each image was separated by its classification

```python
df = pd.read_csv(metadata_path)
labels = df["dx"].unique()
for label in labels:
    os.makedirs(os.path.join(output_path, label), exist_ok=True)
print("Created class folders:", labels)
```

```python
moved = 0
missing = 0
for i, row in df.iterrows():
    img_id = row["image_id"] + ".jpg"
    label = row["dx"]

    src = os.path.join(images_path, img_id)
    dst = os.path.join(output_path, label, img_id)

    if os.path.exists(src):
        shutil.move(src, dst)
        moved += 1
    else:
        missing += 1

print(f"Moved {moved} images.")
print(f"Missing {missing} images (files not found).")
```
